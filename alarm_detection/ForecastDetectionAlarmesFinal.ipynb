# Metodologia
___________________

O **CRISP-DM (Cross Industry Standard Process for Data Mining)** é uma metodologia amplamente utilizada para projetos de ciência de dados e mineração de dados. Ele fornece um processo estruturado dividido em seis fases iterativas, permitindo flexibilidade e adaptação conforme a necessidade do projeto.  
### Etapas do CRISP-DM  
1. **Entendimento do Negócio**: Compreensão do problema, definição dos objetivos e restrições do projeto.  
2. **Entendimento dos Dados**: Coleta, exploração e análise inicial dos dados disponíveis.  
3. **Preparação dos Dados**: Limpeza, transformação e engenharia de atributos para adequação ao modelo.  
4. **Modelagem**: Seleção e treinamento de algoritmos para análise preditiva ou descritiva.  
5. **Avaliação**: Validação dos modelos e alinhamento com os objetivos de negócio.  
6. **Implantação**: Implementação do modelo em ambiente produtivo e monitoramento contínuo.  
### Justificativa para Uso do CRISP-DM  
O CRISP-DM é a abordagem mais indicada para este projeto, pois:  
- **É um padrão consolidado** na indústria, garantindo organização e rastreabilidade do processo.  
- **Promove um ciclo iterativo**, permitindo refinamentos contínuos à medida que novas informações surgem.  
- **Facilita a colaboração** entre cientistas de dados e stakeholders, pois estrutura bem as fases do projeto.  
- **Alinha a solução com o problema de negócio**, garantindo que a modelagem e a análise de dados atendam aos objetivos estratégicos.  
- **Incorpora boas práticas**, como exploração de dados detalhada, preparação rigorosa e validação robusta de modelos antes da implantação.  
Com essa abordagem, garantimos um fluxo de trabalho estruturado, flexível e orientado a resultados, maximizando o impacto do projeto de ciência de dados.

```python
def clean_string(text):
    # Remove acentos convertendo para representação ASCII
    text = unicodedata.normalize('NFKD', text).encode('ASCII', 'ignore').decode('utf-8')
    # Converte para minúsculas
    text = text.lower()
    # Remove caracteres especiais, mantendo apenas letras, números e espaços
    text = re.sub(r'[^a-z0-9\s]', '', text)
    # Remove espaços duplicados e espaços no início/fim
    text = re.sub(r'\s+', ' ', text).strip()
    # Substitui os espaços entre palavras por underscore
    text = text.replace(' ', '_')
    return text

# Aplica a função na coluna 'des_flg_severity'
df_rawData_CSV['des_flg_severity'] = df_rawData_CSV['des_flg_severity'].apply(clean_string)
```

##### Correlacao

Quando trabalhamos com variáveis categóricas (como client_host, des_flg_severity e flg_tipo_item), a noção tradicional de “correlação” (como o coeficiente de Pearson) não se aplica diretamente. Uma forma comum de verificar se há associação entre duas variáveis categóricas é por meio de:

    Tabelas de contingência (crosstab).
    Teste Qui-quadrado (Chi-square).

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Tabela de contingência: linhas = client_host, colunas = severidade
contingency_severity = pd.crosstab(df['client_host'], df['des_flg_severity'])
print("Tabela de contingência (client_host x severidade):\n", contingency_severity)

# Teste Qui-quadrado
chi2, p, dof, expected = chi2_contingency(contingency_severity)
print("\n=== Teste Qui-quadrado (client_host x severidade) ===")
print(f"Estatística qui-quadrado: {chi2}")
print(f"p-valor: {p}")
print(f"Graus de liberdade: {dof}")
print("Frequências esperadas (teóricas):\n", expected)
```

```python
# Tabela de contingência: linhas = client_host, colunas = flg_tipo_item
contingency_item = pd.crosstab(df['client_host'], df['flg_tipo_item'])
print("Tabela de contingência (client_host x tipo_item):\n", contingency_item)

# Teste Qui-quadrado
chi2, p, dof, expected = chi2_contingency(contingency_item)
print("\n=== Teste Qui-quadrado (client_host x tipo_item) ===")
print(f"Estatística qui-quadrado: {chi2}")
print(f"p-valor: {p}")
print(f"Graus de liberdade: {dof}")
print("Frequências esperadas (teóricas):\n", expected)
```

##### Interpretação dos Resultados do Teste Qui-quadrado

Realizado testes de associação (Qui-quadrado) entre:

- `client_host` e `des_flg_severity`
- `client_host` e `flg_tipo_item`

Em ambos os casos, o p-valor (p) foi extremamente baixo (próximo de zero). Esse resultado indica que existe uma associação estatisticamente significativa entre as variáveis analisadas, ou seja, a distribuição de severidades não é independente do `client_host`, e o mesmo vale para a distribuição dos tipos de item.

##### O que isso significa?

- **client_host x des_flg_severity**: Há evidências de que certos hosts (ou clientes) tenham maior propensão a gerar eventos de determinada severidade.  
Por exemplo, alguns hosts podem ter muito mais eventos alto ou desastre em comparação com outros.

- **client_host x flg_tipo_item**: Da mesma forma, a ocorrência de tipos de item (CPU, DISCO, MEMÓRIA etc.) não está distribuída de maneira uniforme entre os hosts.  
Alguns podem concentrar mais eventos de DISCO, enquanto outros podem ter mais de CPU, e assim por diante.

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

# 1. Tabela de contingência (observado)
contingency_severity = pd.crosstab(df['client_host'], df['des_flg_severity'])

# 2. Obter a matriz de frequências esperadas
chi2, p, dof, expected = chi2_contingency(contingency_severity)

# 3. Calcular resíduos padronizados
observed = contingency_severity.values
residuals = (observed - expected) / np.sqrt(expected)

# 4. Transformar em DataFrame para facilitar análise
residuals_df = pd.DataFrame(
    residuals,
    index=contingency_severity.index,
    columns=contingency_severity.columns
)

# 5. Empilhar para termos (client_host, severidade) em linhas
residuals_long = residuals_df.stack().rename("std_resid").reset_index()
residuals_long.rename(columns={"level_0": "client_host", "level_1": "des_flg_severity"}, inplace=True)

# 6. Calcular o valor absoluto para ranquear por maior desvio
residuals_long["abs_std_resid"] = residuals_long["std_resid"].abs()

# 7. Ordenar pelos desvios absolutos (maiores primeiro)
residuals_long.sort_values("abs_std_resid", ascending=False, inplace=True)

# 8. Visualizar o top 10 (por exemplo)
print(residuals_long.head(10))
```

Interpretação do resultado

    Cada linha do residuals_long corresponde a uma combinação (client_host,des_flg_severity)(client_host,des_flg_severity), com o valor do resíduo padronizado em std_resid.
    As primeiras linhas (maiores valores de abs_std_resid) são as “células” que mais se desviam do esperado:
        Se std_resid for positivo e grande, significa que aquele client_host tem mais eventos nessa severidade do que o esperado se não houvesse associação.
        Se std_resid for negativo e grande, significa que aquele client_host tem menos eventos nessa severidade do que o esperado.

Isso retorna um ranking de quais hosts (e quais severidades) têm a associação mais forte, positiva ou negativa.

```python
contingency_item = pd.crosstab(df['client_host'], df['flg_tipo_item'])
chi2, p, dof, expected = chi2_contingency(contingency_item)

observed = contingency_item.values
residuals = (observed - expected) / np.sqrt(expected)

residuals_df = pd.DataFrame(
    residuals,
    index=contingency_item.index,
    columns=contingency_item.columns
)

residuals_long = residuals_df.stack().rename("std_resid").reset_index()
residuals_long.rename(columns={"level_0": "client_host", "level_1": "flg_tipo_item"}, inplace=True)
residuals_long["abs_std_resid"] = residuals_long["std_resid"].abs()
residuals_long.sort_values("abs_std_resid", ascending=False, inplace=True)
```

Interpretação do Resultado

Cada linha do `residuals_long` corresponde a uma combinação (`client_host`, `des_flg_severity`), com o valor do resíduo padronizado em `std_resid`.

As primeiras linhas (maiores valores de `abs_std_resid`) são as “células” que mais se desviam do esperado:
- Se `std_resid` for positivo e grande, significa que aquele `client_host` tem mais eventos nessa severidade do que o esperado se não houvesse associação.
- Se `std_resid` for negativo e grande, significa que aquele `client_host` tem menos eventos nessa severidade do que o esperado.

Isso retorna um ranking de quais hosts (e quais severidades) têm a associação mais forte, positiva ou negativa.

Dessa forma, podemos ranquear quais hosts e categorias se desviam mais fortemente do padrão, o que costuma ser interpretado como “correlação mais forte” no contexto de tabelas de contingência.

### Justificativa para o Uso do Modelo XGBoost

A escolha do modelo XGBoost (Extreme Gradient Boosting) para prever a quantidade de eventos nos hosts monitorados nos próximos 12 meses é fundamentada em diversos fatores que tornam esse algoritmo particularmente eficaz para esse tipo de problema:

1. **Capacidade de Capturar Relações Não Lineares**  
   O conjunto de dados contém múltiplas variáveis categóricas e temporais que podem influenciar a frequência dos eventos de maneira não linear. O XGBoost, sendo um modelo baseado em árvores de decisão, é altamente eficaz para capturar essas interações complexas sem exigir transformações explícitas das variáveis.

2. **Manejo Eficiente de Dados Categóricos**  
   Como observado na estrutura da base de dados, há diversas colunas categóricas, como **severidade**, **cliente**, **item**, **host**, entre outras. O XGBoost consegue lidar bem com essas variáveis, especialmente após a aplicação de técnicas de codificação adequadas (como Target Encoding ou One-Hot Encoding).

3. **Robustez contra Overfitting**  
   A implementação do XGBoost possui regularização (L1 e L2), o que reduz o risco de overfitting, um fator crítico considerando que a previsão abrange um horizonte de 12 meses, exigindo que o modelo generalize bem para dados futuros.

4. **Eficácia em Séries Temporais via Modelagem Direta**  
   Apesar de XGBoost não ser um modelo de séries temporais nativo como ARIMA ou LSTM, ele pode ser adaptado para esse tipo de problema através da criação de features temporais (lags, médias móveis, tendências sazonais, etc.). Esse método permite capturar padrões históricos e projetá-los para o futuro.

5. **Escalabilidade e Desempenho Computacional**  
   O XGBoost é conhecido por sua eficiência computacional, utilizando otimizações como paralelismo e processamento em árvore binária comprimida. Isso o torna uma escolha viável para bases grandes, como a utilizada neste projeto (com potencial de crescimento para mais de 200.000 registros).

6. **Capacidade de Explicabilidade**  
   Diferente de redes neurais, que atuam como “caixas-pretas”, o XGBoost permite interpretar a importância das features através de métodos como SHAP (Shapley Additive Explanations) e análise de ganho nas árvores. Isso possibilita maior compreensão dos fatores que impactam a previsão de eventos.

7. **Histórico de Sucesso em Aplicações Similares**  
   Modelos baseados em Gradient Boosting, como o XGBoost, são amplamente utilizados em problemas de previsão de demanda, detecção de falhas e monitoramento de sistemas devido à sua flexibilidade e capacidade preditiva. Sua eficácia já foi comprovada em diversos cenários de previsão.

O **XGBoost se mostra uma escolha apropriada para este problema devido à sua robustez, capacidade de capturar padrões complexos e bom desempenho computacional.** No entanto, para garantir uma melhor acurácia, serão exploradas técnicas de ajuste de hiperparâmetros, seleção de features e validação cruzada.

Em seguida, criamos novas features de tempo, incluindo o ano, mês e um índice temporal 't', que nos permitirá realizar análises mais detalhadas ao longo do tempo. Essas features são essenciais para identificar tendências e padrões no comportamento dos itens de CPU ao longo dos períodos analisados.

Vamos prosseguir com a análise exploratória dos dados e identificar insights relevantes que possam ajudar na tomada de decisões estratégicas.

```python
# Filtrar o grupo de exemplo
group_example = df[
    (df['des_flg_severity'] == 'alto') &
    (df['client_host'] == '14_10634') &
    (df['flg_tipo_item'] == 'CPU')
].copy()

# Ordenar pela coluna de data
group_example = group_example.sort_values('ano_mes')

# Criar features de tempo: ano, mês e um índice temporal 't'
group_example['year'] = group_example['ano_mes'].dt.year
group_example['month'] = group_example['ano_mes'].dt.month
group_example['t'] = range(1, len(group_example) + 1)
```

Agora, vamos criar uma feature de defasagem (lag) para a variável qtde, a fim de capturar a dependência temporal. Para isso, adicionaremos uma nova coluna que representará o valor de qtde do mês anterior.

**Nota Explicativa: Uso de Coluna de Defasagem (Lag)**

A criação de uma coluna de defasagem (lag) para a variável `qtde` é uma técnica utilizada para capturar a dependência temporal dos dados.  
Ao incluir o valor de `qtde` do mês anterior, podemos analisar como os valores passados influenciam os valores futuros. Isso é particularmente útil em modelos de previsão, onde a identificação de padrões temporais pode melhorar a precisão das previsões.  
A coluna de defasagem nos permite incorporar essa informação temporal diretamente no modelo, facilitando a análise e a modelagem dos dados.

```python
# Criar a feature de defasagem para qtde
group_example['qtde_lag1'] = group_example['qtde'].shift(1)
group_example = group_example.dropna().reset_index(drop=True)

print(group_example.head())
```

Em seguida, procederemos com a preparação dos dados para o XGBoost, definindo as features e o target, e realizando a divisão dos dados em conjuntos de treino e teste.

#### Treino do modelo

Agora, vamos preparar os dados para treinar e validar um modelo XGBoost utilizando as features criadas.  
Vamos dividir os dados em **treino e teste (80% / 20%)**. A divisão é feita com base na ordem dos dados, onde os primeiros 80% são usados para treino e os últimos 20% para teste.  
Essa abordagem garante que os dados de treino e teste respeitem a **sequência temporal, o que é importante para modelos de séries temporais**.

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_percentage_error

# Definir as features e o target
features = ['t', 'year', 'month', 'qtde_lag1']
X = group_example[features]
y = group_example['qtde']

# Dividir os dados em treino e teste (80% treino, 20% teste)
train_size = int(len(group_example) * 0.8)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

# Inicializar e treinar o modelo XGBoost
model = XGBRegressor(objective='reg:squarederror', random_state=42)
model.fit(X_train, y_train)

# Previsões no conjunto de teste
y_pred = model.predict(X_test)

# Cálculo das métricas RMSE e MAPE
rmse = mean_squared_error(y_test, y_pred, squared=False)
mape = mean_absolute_percentage_error(y_test, y_pred)

print("RMSE:", rmse)
print("MAPE:", mape)
```

### Métricas


RMSE (Root Mean Square Error): RMSE mede a diferença entre os valores previstos e os reais. Calcula a média das diferenças ao quadrado e tira a raiz quadrada dessa média. Menor RMSE indica melhor precisão. Imagine que você está tentando prever a altura de várias pessoas. O RMSE calcula a média das diferenças ao quadrado entre as alturas previstas e as alturas reais, e depois tira a raiz quadrada dessa média. Quanto menor o RMSE, melhor o modelo está prevendo.

MAPE (Mean Absolute Percentage Error): MAPE mede a precisão das previsões em porcentagem. Calcula a média das diferenças absolutas dividida pelos valores reais. Menor MAPE indica melhor precisão. Por exemplo, se você está prevendo vendas e o MAPE é 10%, isso significa que, em média, suas previsões estão erradas por 10% do valor real.

Importância em Séries Temporais Multivariadas:

RMSE: Avalia a precisão global considerando todas as variáveis e períodos. Sensível a grandes erros, útil para identificar previsões fora do esperado.
MAPE: Facilita a interpretação em porcentagem, permitindo comparar a precisão entre variáveis de diferentes escalas.
Resumo:

RMSE: Precisão global e sensível a grandes erros.  
MAPE: Interpretação intuitiva e comparação entre diferentes escalas.

####  Previsões Iterativas

Agora que temos um modelo treinado para o grupo de exemplo, o próximo passo é utilizar esse modelo para fazer previsões iterativas para os próximos 11 meses. O procedimento será o seguinte:

1. Obter o último registro do grupo, que servirá de base para a primeira previsão.
2. Para cada mês seguinte, gerar a nova data, atualizar o índice temporal (t) e as variáveis de ano e mês.
3. Utilizar o valor previsto do mês anterior como defasagem para a próxima previsão (`qtde_lag1`).
4. Armazenar os resultados em um DataFrame.

Esse código realizará previsões para os próximos 11 meses de forma iterativa, utilizando o valor previsto como defasagem para a próxima etapa. Por favor, execute essa célula e compartilhe os resultados para prosseguirmos com a análise.

```python
# Obter o último registro do grupo
last_record = group_example.iloc[-1]

# Inicializar variáveis a partir do último registro
current_date = last_record['ano_mes']
current_t = last_record['t']
current_qtde_lag1 = last_record['qtde']  # para a primeira previsão

forecast_steps = 11
forecast_results = []

for i in range(forecast_steps):
    # Atualiza a data: adiciona 1 mês
    new_date = current_date + relativedelta(months=1)
    new_t = current_t + 1
    new_year = new_date.year
    new_month = new_date.month
    
    # Preparar o DataFrame para previsão
    X_new = pd.DataFrame({
        't': [new_t],
        'year': [new_year],
        'month': [new_month],
        'qtde_lag1': [current_qtde_lag1]
    })
    
    # Previsão para o mês atual
    pred = model.predict(X_new)[0]
    
    forecast_results.append({
        'ano_mes': new_date,
        'pred_qtde': pred,
        't': new_t,
        'year': new_year,
        'month': new_month,
        'qtde_lag1': current_qtde_lag1
    })
    
    # Atualiza as variáveis para a próxima iteração
    current_date = new_date
    current_t = new_t
    current_qtde_lag1 = pred  # utilizar o valor previsto como lag para o próximo mês

forecast_df = pd.DataFrame(forecast_results)
print(forecast_df)
```

Agora que conseguimos realizar a previsão iterativa para um grupo de exemplo, vamos encapsular esse processo em uma função para facilitar a aplicação em outros grupos. Dessa forma, poderemos aplicar a mesma lógica para cada combinação de des_flg_severity, client_host e flg_tipo_item.

```python
from dateutil.relativedelta import relativedelta
from xgboost import XGBRegressor
import pandas as pd

def forecast_group(df_group, forecast_steps=11):
    # Ordena os dados e recria as features temporais
    df_group = df_group.sort_values('ano_mes').copy()
    df_group['year'] = df_group['ano_mes'].dt.year
    df_group['month'] = df_group['ano_mes'].dt.month
    df_group['t'] = range(1, len(df_group) + 1)
    
    # Cria a feature de defasagem e remove o primeiro registro (NaN)
    df_group['qtde_lag1'] = df_group['qtde'].shift(1)
    df_group = df_group.dropna().reset_index(drop=True)
    
    # Se não houver registros suficientes, retorna None
    if df_group.shape[0] < 2:
        return None
    
    # Define as features e o target
    features = ['t', 'year', 'month', 'qtde_lag1']
    X = df_group[features]
    y = df_group['qtde']
    
    # Treina o modelo XGBoost com todos os dados do grupo
    model = XGBRegressor(objective='reg:squarederror', random_state=42)
    model.fit(X, y)
    
    # Obtém o último registro para iniciar as previsões
    last_record = df_group.iloc[-1]
    current_date = last_record['ano_mes']
    current_t = last_record['t']
    current_qtde_lag1 = last_record['qtde']
    
    # Realiza as previsões iterativas para os próximos 'forecast_steps' meses
    forecast_results = []
    for i in range(forecast_steps):
        new_date = current_date + relativedelta(months=1)
        new_t = current_t + 1
        new_year = new_date.year
        new_month = new_date.month
        
        # Prepara as features para o novo mês
        X_new = pd.DataFrame({
            't': [new_t],
            'year': [new_year],
            'month': [new_month],
            'qtde_lag1': [current_qtde_lag1]
        })
        
        # Realiza a previsão
        pred = model.predict(X_new)[0]
        forecast_results.append({
            'ano_mes': new_date,
            'pred_qtde': pred
        })
        
        # Atualiza as variáveis para a próxima iteração
        current_date = new_date
        current_t = new_t
        current_qtde_lag1 = pred
    
    return pd.DataFrame(forecast_results)

# Teste da função com o grupo de exemplo:
example_group = df[
    (df['des_flg_severity'] == 'alto') &
    (df['client_host'] == '14_10634') &
    (df['flg_tipo_item'] == 'CPU')
]

forecast_example = forecast_group(example_group)
print(forecast_example)
```

Agora que temos uma função que gera a previsão para um grupo, vamos aplicar essa função para cada combinação única de des_flg_severity, client_host e flg_tipo_item. Dessa forma, conseguiremos um DataFrame com as previsões para os próximos 11 meses para cada grupo.



Assim, podemos analisar se as previsões foram geradas conforme o esperado e discutir os próximos passos, como por exemplo, agregações ou análises adicionais dos resultados.

### Previsoes

```python
# Definir as colunas que identificam cada grupo
group_columns = ['des_flg_severity', 'client_host', 'flg_tipo_item']

# Lista para armazenar os resultados de previsão de cada grupo
forecast_all = []

# Loop sobre cada grupo
for key, group in df.groupby(group_columns):
    forecast = forecast_group(group, forecast_steps=11)
    if forecast is not None:
        # Acrescenta as informações do grupo no DataFrame de forecast
        for col, value in zip(group_columns, key):
            forecast[col] = value
        forecast_all.append(forecast)

# Concatena todas as previsões em um único DataFrame
forecast_all_df = pd.concat(forecast_all, ignore_index=True)
print(forecast_all_df.head())
```

```python
from dateutil.relativedelta import relativedelta
from xgboost import XGBRegressor
import pandas as pd

def forecast_group(df_group, forecast_steps=12):
    # Ordena e recria as features temporais
    df_group = df_group.sort_values('ano_mes').copy()
    df_group['year'] = df_group['ano_mes'].dt.year
    df_group['month'] = df_group['ano_mes'].dt.month
    df_group['t'] = range(1, len(df_group) + 1)
    
    # Cria a feature de defasagem e remove o primeiro registro
    df_group['qtde_lag1'] = df_group['qtde'].shift(1)
    df_group = df_group.dropna().reset_index(drop=True)
    
    # Se o grupo não tiver registros suficientes, retorna None
    if df_group.shape[0] < 2:
        return None
    
    # Define as features e o target
    features = ['t', 'year', 'month', 'qtde_lag1']
    X = df_group[features]
    y = df_group['qtde']
    
    # Treina o modelo XGBoost com os dados disponíveis do grupo
    model = XGBRegressor(objective='reg:squarederror', random_state=42)
    model.fit(X, y)
    
    # Obtém o último registro para iniciar as previsões
    last_record = df_group.iloc[-1]
    current_date = last_record['ano_mes']
    current_t = last_record['t']
    current_qtde_lag1 = last_record['qtde']
    
    # Realiza as previsões para os próximos 12 meses
    forecast_results = []
    for i in range(forecast_steps):
        new_date = current_date + relativedelta(months=1)
        new_t = current_t + 1
        new_year = new_date.year
        new_month = new_date.month
        forecast_period = i + 1  # 1 para o primeiro mês, 2 para o segundo, etc.
        
        # Prepara as features para previsão
        X_new = pd.DataFrame({
            't': [new_t],
            'year': [new_year],
            'month': [new_month],
            'qtde_lag1': [current_qtde_lag1]
        })
        pred = model.predict(X_new)[0]
        
        forecast_results.append({
            'ano_mes': new_date,
            'forecast_period': forecast_period,
            'pred_qtde': pred
        })
        
        # Atualiza as variáveis para a próxima iteração
        current_date = new_date
        current_t = new_t
        current_qtde_lag1 = pred
        
    return pd.DataFrame(forecast_results)

# Teste com o grupo de exemplo (por exemplo, severidade "alto", client_host "14_10634" e item "CPU")
example_group = df[
    (df['des_flg_severity'] == 'alto') &
    (df['client_host'] == '14_10634') &
    (df['flg_tipo_item'] == 'CPU')
]
forecast_example = forecast_group(example_group)
print(forecast_example)
```

```python
# Definir as colunas que identificam cada grupo
group_columns = ['des_flg_severity', 'client_host', 'flg_tipo_item']

# Lista para armazenar as previsões de cada grupo
forecast_all = []

# Loop em cada grupo
for key, group in df.groupby(group_columns):
    forecast = forecast_group(group, forecast_steps=12)
    if forecast is not None:
        # Acrescenta as informações do grupo à previsão
        for col, value in zip(group_columns, key):
            forecast[col] = value
        forecast_all.append(forecast)

# Concatena todas as previsões em um único DataFrame
forecast_all_df = pd.concat(forecast_all, ignore_index=True)
print(forecast_all_df.head(10))
```
